# Offshore Environment Setup
Run this cell to setup the environment in Google Colab or Kaggle. It will clone the repository, install dependencies, and configure persistent storage.

```python
import os
import sys
import subprocess

# 1. Environment Detection
IN_COLAB = 'google.colab' in sys.modules
IN_KAGGLE = 'KAGGLE_KERNEL_RUN_TYPE' in os.environ

repo_url = "https://github.com/your-username/offshore.git" # Replace with actual repo URL if private/different
repo_dir = "offshore"

# 2. Clone Repository
if not os.path.exists(repo_dir):
    print("Cloning repository...")
    subprocess.run(["git", "clone", repo_url])
else:
    print("Repository already exists, pulling latest...")
    subprocess.run(["git", "-C", repo_dir, "pull"])

# 3. Set Working Directory
os.chdir(repo_dir)
print(f"Working directory set to: {os.getcwd()}")

# 4. Install Dependencies
print("Installing dependencies (this may take a minute)...")
subprocess.run([sys.executable, "-m", "pip", "install", "-r", "requirements.txt"], check=True)

# 5. Configure Persistent Storage
if IN_COLAB:
    from google.colab import drive
    print("Mounting Google Drive for persistent storage...")
    drive.mount('/content/drive')
    # Set output directories to Google Drive
    os.environ["OFFSHORE_MODEL_DIR"] = "/content/drive/MyDrive/offshore/models/weights"
    os.environ["OFFSHORE_MLOPS_DIR"] = "/content/drive/MyDrive/offshore/mlops"
    print("Configured to save artifacts to Google Drive (/content/drive/MyDrive/offshore/)")
elif IN_KAGGLE:
    # Set output directories to Kaggle's persistent working directory
    os.environ["OFFSHORE_MODEL_DIR"] = "/kaggle/working/offshore/models/weights"
    os.environ["OFFSHORE_MLOPS_DIR"] = "/kaggle/working/offshore/mlops"
    print("Configured to save artifacts to /kaggle/working/offshore/")
else:
    print("Running locally, using default relative paths.")

# 6. Verify GPU
try:
    import torch
    if torch.cuda.is_available():
        print(f"\n\u2705 GPU Detected: {torch.cuda.get_device_name(0)}")
    else:
        print("\n\u26a0\ufe0f No GPU detected - running on CPU")
except ImportError:
    print("\nPyTorch not installed correctly.")
```
